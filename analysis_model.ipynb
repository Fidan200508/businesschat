```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=1)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.model_selection import GridSearchCV
```

```python
def train_model(model, param_grid,X_train, y_train, X_test, y_test):
    grid=GridSearchCV(model, param_grid, cv=5, scoring='accuracy', n_jobs=-1)
    grid.fit(X_train, y_train)
    base_model=grid.best_estimator_
    y_pred=base_model.predict(X_test)
    acc=accuracy_score(y_test, y_pred)
    return base_model, acc, grid.best_params_
```

```python
dt = DecisionTreeClassifier(random_state=1, class_weight='balanced')
dt_param_grid = {
    'max_depth': [3,5,7,10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}
```

```python
best_dt, acc_dt, params_dt = train_model(dt, dt_param_grid, X_train, y_train, X_test, y_test)
```

```python
print("Decision Tree Accuracy:", acc_dt)
```

```python
print("Best parameters for Decision Tree:", params_dt)
```

```python
from imblearn.over_sampling import SMOTE

# 1. Define the SMOTE object
sm = SMOTE(random_state=1)

# 2. Fit and apply it ONLY to your training data
# Note: Never apply SMOTE to your test set!
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)

# 3. Now pass X_train_res and y_train_res to your train_model function
best_dt, acc_dt, params_dt = train_model(dt, dt_param_grid, X_train_res, y_train_res, X_test, y_test)
```

```python
from imblearn.over_sampling import SMOTE

# 1. Initialize SMOTE
sm = SMOTE(random_state=1)

# 2. Balance the training data
# We only resample the training set so the testing set remains 100% real data
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)

# 3. Verify the balance
import pandas as pd
print("New Class Distribution:\n", pd.Series(y_train_res).value_counts())
```
